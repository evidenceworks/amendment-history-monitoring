# Quick check

This verifies package integrity, compact source identity, headline reference results and Figures 1–4. It does not fit models, run R, or reconstruct the raw source.


```python
from pathlib import Path
import json, platform, subprocess, time
from google.colab import files
START = time.perf_counter()
ROOT = Path("/content/amendment-history-monitoring")
if not ROOT.is_dir():
    subprocess.run(["git", "clone", "--depth", "1", "https://github.com/evidenceworks/amendment-history-monitoring.git", str(ROOT)], check=True)
print({"repository": str(ROOT)})
```

```python
quick_receipt = Path("/content/quick_check_receipt.json")
subprocess.run(["python3", str(ROOT/"verification/verify_science.py")], cwd=ROOT, check=True)
subprocess.run(["python3", str(ROOT/"scripts/quick_check.py"), "--json-out", str(quick_receipt)], cwd=ROOT, check=True)
report = json.loads(quick_receipt.read_text())
report.update({"transport":"repository clone or existing checkout","total_seconds":time.perf_counter()-START,"platform":platform.platform()})
if report["total_seconds"] > 120: raise RuntimeError(f"Quick runtime ceiling exceeded: {report['total_seconds']:.3f}s")
quick_receipt.write_text(json.dumps(report,indent=2)+"\n")
print(json.dumps(report,indent=2))
from IPython.display import display, Image
display(Image(filename=str(ROOT/"figures/final/figure4.png")))
```
